# union


```python
# now create two set
python_students={"rahul",'vani',"manish","anil","chirag"}
java_students={"manish","anil","llith","sanjiv"}

#union operation
result=python_students.union(java_students)
print("union operations: ",result)

#intersection
result_inter=python_students.intersection(java_students)
print("intersection operations: ",result_inter)

#difference
result_diff=python_students.difference(java_students)
print("difference operations: ",result_diff)

#diffrence
res_diff=python_students.difference(java_students)
print("difference operations: ",res_diff)
```
