```python
#create 2 table
Students_info=pd.DataFrame({
    "Students ID":[ "S101", "S1301", "S1014", "S8101", "S1031",
        "S1071", "S31041", "S181", "S103"],
    "Name":[        "Anu", "Ravi", "Priya", "Arun", "Kamal",
        "Meena", "Vijay", "Neha", "Rahul"],
    "Department":[        "CSE", "ECE", "CSE", "ECE", "CSE",
        "ECE", "CSE", "ECE", "CSE"],
})
print(Students_info)
```

```python
import pandas as pd

# 2. Examination Table
exam = pd.DataFrame({
    "Students_ID": [
        "S101", "S1301", "S1014", "S8101", "S1031",
        "S1071", "S31041", "S181", "S103"
    ],
    "Semester": [2, 3, 4, 2, 5, 3, 4, 1, 6],
    "Final": [82, 75, 94, 80, 68, 85, 78, 90, 88]
})

print(exam)
```

```python
# Merge
# Since 'Students_info' has 'Students ID' (with a space) and 'exam' has 'Students_ID' (with an underscore),
# we specify left_on and right_on parameters to join them correctly.
result = pd.merge(Students_info, exam, left_on="Students ID", right_on="Students_ID", how="inner")
print(result)
```

```python
#Ljoin
Leftjoin=pd.merge(Students_info,exam,left_on="Students ID",right_on="Students_ID",how="left")
print(Leftjoin)
```

```python
    # Right Join
Rightjoin = pd.merge(Students_info, exam, left_on="Students ID", right_on="Students_ID", how="right")
print(Rightjoin)
```

```python
# Outer Join
Outerjoin = pd.merge(Students_info, exam, left_on="Students ID", right_on="Students_ID", how="outer")
print(Outerjoin)
```

```python
# Concatenation
import pandas as pd
batch1=pd.DataFrame({
    "Students_ID":["S101","S102"],
    "Marks":[88,72]

})
batch2=pd.DataFrame({
    "Students_ID":["S103","S104"],
    "Marks":[94,80]
})
all_students=pd.concat([batch1,batch2])
print(all_students)
```

```python
#axis =0 axis=1
a1=pd.concat([batch1,batch2],axis=1)
print(a1)
```

```python
#axis =0 axis=1
a2=pd.concat([batch1,batch2],axis=0,ignore_index=True)
print(all_students)
```

```python
#Reshaping and Pivot
#Changing Rows Into Columns Or Rows In to columns
s1=pd.DataFrame({
    "Students_ID":["S101","S102","S103"],
    "Python":[88,72,77],
    "Statistics":[56,79,44],
    "ML":[67,87,34]

})
s1
```

```python
import pandas as pd

data = {
    "Students_ID": [101, 102, 103],
    "Maths": [85, 90, 78],
    "Science": [88, 76, 92]
}

wide = pd.DataFrame(data)

print(wide)
```

```python
long_result = wide.melt(
    id_vars="Students_ID",
    var_name="Subject",
    value_name="marks"
)

print(long_result)
```

```python
#pivot_table
pivot_table_agg=pd.pivot_table(long_result,index="Students_ID",columns="Subject",values="marks",aggfunc="mean")
print(pivot_table_agg)
```
