```python
import pandas as pd
```

```python
raw_data={
    "Name":["Amit","Riya","Priya","Karan","Rahul"],
    "Age":[20 , 21 , 20 , 21 , 22],
    "Department":["CSE","IT","ECE","CSE","IT"],
    "Marks":[85,90,78,88,76]
}
df=pd.DataFrame(raw_data)
df
```

```python

# Inspect DataFrame

print("Shape (rows, columns):", df.shape)
print(df.head(3))
print(df.tail(2))
print(df.columns)
print(df.columns.tolist())
print(df.dtypes)
print(df.describe())
df.info()
```

```python
iris=pd.read_csv("/content/Iris.csv")
iris.head(3)
```

```python
print(iris.shape)
print(iris.columns.tolist())
print(iris.info())
```

```python
# file formats
iris.to_csv("iris_csvoutput.csv",index=False)
#load
iris_csv=pd.read_csv("/content/iris_csvoutput.csv")
iris_csv.head(7)
```

```python
# Save the Iris DataFrame to an Excel file
iris.to_excel("iris_exceloutput.xlsx", index=False)

# Load the Excel file
import pandas as pd

iris_excel = pd.read_excel("iris_exceloutput.xlsx")
iris_excel.tail(8)
```

```python
iris.to_json("iris_joutput.json",orient="records",
             indent=4)
#load and execute it
iris_json=pd.read_json("/content/iris_joutput.json")
iris_json.head(6)
```

```python
iris.to_parquet("iris_parquet.parquet",index=False)
#load and display
iris_parquet=pd.read_parquet("/content/iris_parquet.parquet")
iris_parquet.head(6)
```

```python
import os
files=[
    "iris_csvoutput.csv",
    "iris_exceloutput.xlsx",
    "iris_joutput.json",
    "iris_parquet.parquet"
]
for file in files:
  size=os.path.getsize(file)
  print(f"{file}:{size} bytes")
```
